date formatting

```python
  parameter: date_filter {
    label: "Time Period"
    view_label: "Component Filters"
    type: unquoted
    allowed_value: {
      label: "Yesterday"
      value: "day"
    }
    allowed_value: {
      label: "Previous Week"
      value: "week"
    }
    allowed_value: {
      label: "Previous Month"
      value: "month"
    }
    allowed_value: {
      label: "6 Months"
      value: "sixmonth"
    }
```

```python
  dimension: date_format {
    type: string
    hidden: yes
    sql:
        {% if date_filter._is_filtered %}
              {% if date_filter._parameter_value == 'sixmonth' %} 'sixmonth'
              {% elsif date_filter._parameter_value == 'month' %} 'month'
              {% elsif date_filter._parameter_value == 'week' %} 'week'
              {% elsif date_filter._parameter_value == 'day' %} 'day'
              {% else %} 'DAY'
              {% endif %}
            {% else %} NULL
            {% endif %}
;;}
```

```python
  dimension: filtered_date {
    label: "Date Range"
    view_label: "0) All Traffic"
    type: date_time
    description: "Displays the specific dates within the date range"
    sql:
      CASE
      WHEN ${date_format} = 'day' THEN date_trunc('hour', ${started_raw})
      WHEN ${date_format} = 'week' THEN date_trunc('day', ${started_raw})
      WHEN ${date_format} = 'month' THEN date_trunc('day', ${started_raw})
      WHEN ${date_format} = 'sixmonth' THEN date_trunc('week', ${started_raw})
      ELSE 'Contact the data team' END
                                ;;
    can_filter: no
  }
```
